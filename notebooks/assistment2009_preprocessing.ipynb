```python
## https://sites.google.com/site/assistmentsdata/home/2009-2010-assistment-data/skill-builder-data-2009-2010?authuser=0

import pandas as pd

skill_builder = pd.read_csv('skill_builder_data.csv', encoding='latin1')

# 필요한 칼럼만 선택
skill_builder = skill_builder[['user_id','order_id','problem_id','correct','skill_id', 'skill_name']]

# 결측치 행 제거
skill_builder = skill_builder.dropna().sort_values('order_id')

# 중복된 데이터 확인
mask = skill_builder.duplicated(subset=['user_id', 'problem_id'], keep=False)

# 한 문제가 여러 개념을 가진 경우... 한 건의 문제풀이를 여러 행으로 분리
# 문항 단위로 예측한다면 성능 조작 가능성?
skill_builder[mask]
```

```python
skill_builder['skill_ID'] = skill_builder['skill_id']

# 필터링: 너무 적은 데이터는 제외
skill_builder = skill_builder.groupby('user_id').filter(lambda q: len(q) > 5).copy()
skill_builder = skill_builder.groupby('skill_id').filter(lambda q: len(q) > 1).copy()

# 인코딩
skill_builder["skill_id"], skill_labels = pd.factorize(skill_builder["skill_id"])
skill_builder["skill_id"] += 1  

skill_builder['correct'] = skill_builder['correct'].astype(int)

skill_builder.to_csv('assistment2009_processsed.csv', encoding='utf-8-sig', index=False)
```
