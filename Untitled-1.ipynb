```python
import json
import pandas as pd

with open("data/raw/[라벨]수학 지식체계 데이터 세트_210611.json", "r", encoding="utf-8") as f:
    kc_map = json.load(f)
```

```python
kc_map['0']
```

```python
from_id = []
to_id = []
for key, value in kc_map.items():
    from_concept = value["fromConcept"]
    to_concept = value["toConcept"]
    
    from_id.append({
        "kc_id": from_concept["id"],
        "kc_name": from_concept["name"],
        "semester": from_concept["semester"],
        "description": from_concept["description"],
        "chapter": from_concept["chapter"]["name"],
        "achievement": from_concept["achievement"]["name"]
    })

    to_id.append({    
        "kc_id": to_concept["id"],
        "kc_name": to_concept["name"],
        "semester": to_concept["semester"],
        "description": to_concept["description"],
        "chapter": to_concept["chapter"]["name"],
        "achievement": to_concept["achievement"]["name"]
    })

from_id = pd.DataFrame(from_id)
to_id = pd.DataFrame(to_id)

kc_map_df = pd.concat([from_id, to_id])

kc_map_df = kc_map_df.drop_duplicates().reset_index(drop=True)
```

```python
from_id = []
to_id = []

for key, value in kc_map.items(): 
    from_id.append(value["fromConcept"]['id'])
    to_id.append(value["toConcept"]['id'])

from_id = pd.DataFrame(from_id, columns=['from_id'])
to_id = pd.DataFrame(to_id, columns=['to_id'])

kc_id_map = pd.concat([from_id, to_id], axis=1)
```

```python
from dotenv import load_dotenv
from neo4j import GraphDatabase
import os

load_dotenv()

URI = os.getenv("NEO4J_URI")
USERNAME = os.getenv("NEO4J_USER")
PASSWORD = os.getenv("NEO4J_PASSWORD")

driver = GraphDatabase.driver(URI, auth=(USERNAME, PASSWORD))

# Test connection
try:
    driver.verify_connectivity()
    print("✅ Connected successfully!")
except Exception as e:
    print("❌ Connection failed:", e)
finally:
    driver.close()
```
